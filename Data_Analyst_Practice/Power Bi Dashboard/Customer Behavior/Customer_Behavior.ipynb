```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df.columns = df.columns.str.replace(' ', '_').str.lower()
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
df.columns
```

```python
# create a column age_group
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
df[['frequency_of_purchases']].head(10)
```

```python
df_group_frequency = df.groupby(['frequency_of_purchases']).count().reset_index()[['frequency_of_purchases']]
df_group_frequency
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90   
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
df[['discount_applied', 'promo_code_used']]
```

```python
# Returns a single True or False
is_identical = df['discount_applied'].equals(df['promo_code_used'])
print(is_identical)
```

```python
df = df.drop('promo_code_used', axis=1)
```

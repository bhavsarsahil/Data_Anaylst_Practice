```python
import pandas as pd

data = {
    "Months" : ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],
}
df = pd.DataFrame(data)
df

df["Short Month"] = df["Months"].str[:3]

# alternative way to create short month column
```

```python
data1 = {
    "Months" : ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],
}

df1 = pd.DataFrame(data1)
# print(df1)
def short_month(month):
    return month[:3]

df1["Short Month"] = df1["Months"].map(short_month)
df1
```

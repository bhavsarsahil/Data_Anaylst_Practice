```python
import pandas as pd
```

```python
df = pd.read_csv("employee_data.csv")
df
```

```python
df.isnull().sum()
```

```python
# if we have some thing not neccesary data we can drop it with using dropna() method
# df.dropna(aixs=0, inplace=True) # axis=0 means drop rows with missing values
# df.dropna(axis=1, inplace=True) # axis=1 means drop columns with missing values
# df.dropna(inplace=True) # default is axis=0

# also we have fillna() method to fill missing values with some specific value
# df.fillna(0, inplace=True) # fill missing values with 0
```

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
df
```

```python
df["Name"] = df["Name"].fillna("Unknown")
df
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
df
```

```python
# sorting data 
# df.sort_values(by = "column_name", ascending=True/False, inplace=True)

df.sort_values(by="Age", ascending=False, inplace=True)
df
```

```python
df.sort_values(by=["Age", "Salary"], ascending=[False, False], inplace=True)
df
```

```python
df["Age"] = df["Age"].astype(int)
df
```

```python
grouped = df.groupby("Age")["Salary"].sum()
grouped
```

```python
# multy grouping
grouped = df.groupby(["Age","Name"])["Salary"].sum()
grouped 

```

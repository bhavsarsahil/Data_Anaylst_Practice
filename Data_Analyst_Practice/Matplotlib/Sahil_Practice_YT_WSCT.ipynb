```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

# Box Plot

```python
l = np.random.randint(1, 100, 30)
describe = pd.Series(l).describe()
print(describe)

print("Original Array or List \n")
print(l)

sorted_l = sorted(l)
print("Sorted Array or List \n")
print(sorted_l, "\n")

split1 = np.array_split(sorted_l,2)
print("Spliting the array into 2 parts \n")
print(split1)

print("Middle elements using array_split : \n")
reslt = split1[0][-1]

print("Middle elements using array_split: ", reslt)
# l = [25,46,78,34,56,75,23,45,67,89]
plt.boxplot(l)
plt.show()
```

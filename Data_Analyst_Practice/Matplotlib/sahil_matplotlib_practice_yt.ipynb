```python
product = ["A", "B", "C", "D"]
sales = [1000, 1500, 800, 1200]
# use height (bar thickness) and not a comparison; for horizontal bars set x limits (values on x-axis)
plt.barh(product, sales, color="orange", height=0.5, label="Product 2025")
plt.title("Product Sales Data")
plt.xlabel("Sales")
plt.ylabel("Products")
plt.xlim(0, 1600)
plt.legend()
plt.show()
```

# Subplot in Matploatlib

```python
#plt.subplot(nrows, ncols, index)
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(1,2,1) # 1st row , 2nd column, 1st subplot
plt.plot(x, y, color="blue", marker="o")
plt.title("Line Plot")

plt.subplot(1,2,2) # 1st row , 2nd column, 2nd subplot
plt.bar(x, y, color="green")
plt.title("Bar Chart")

# plt.tight_layout()
plt.show()
```

```python
fig,ax = plt.subplots(1,2 , figsize=(10,5)) # (width, height)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

ax[0].plot(x, y, color="blue", marker="o")
ax[0].set_title("Line Plot")

ax[1].bar(x, y, color="green")
ax[1].set_title("Bar Chart")

fig.suptitle("Sales Data Visualization", fontsize=16)

plt.tight_layout()
plt.show()
```

# Project On Netflix DataSet

```python
import pandas as pd
import matplotlib.pyplot as plt

# load the dataset

df = pd.read_csv("netflix_titles.csv")
df
```

```python
df.isnull().sum()
```

```python
type_counts = df['type'].value_counts()
type_counts
```

```python
plt.bar(type_counts.index, type_counts.values, color=["skyblue", "orange"])
plt.title("Distribution of Content Types on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Count")
plt.figure(figsize=(6,4))
plt.tight_layout()
plt.show()
```

```python
rating_counts = df['rating'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(rating_counts.values, labels=rating_counts.index, autopct="%.1f%%", startangle=140, shadow=True)
plt.title("Distribution of Content Ratings on Netflix") 
plt.tight_layout()
plt.show()
```

```python
movie_df = df[df['type'] == 'Movie'].copy()
# Remove rows where 'duration' is NaN before conversion
movie_df = movie_df[movie_df['duration'].notna()]
movie_df['duration_int'] = movie_df['duration'].str.replace(' min', '').astype(int)

plt.figure(figsize=(8,6))
plt.hist(movie_df['duration_int'], bins=20, color="teal", edgecolor="black", alpha=0.7)
plt.title("Distribution of Movie Durations on Netflix")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```

# Matplotlib :

## What is matplotilb ?
* mateplotilb is a low level graph plotting library in python that seves as visualization utility
* matplotib is open source and can use it freely.
* matplotib is powerful plotting library in python used for creating static,animated and interactive visualizations.Matplotib's primary purpose is to provide users with the tools and functionality to represent data graphically,make it easier to analyze and understand
* it was originally developed by johh D. Hunter in 2003 and is now maintained by a large communty of developers.

```python
# Import Matplotlib For  visualization.
import matplotlib.pyplot as plt
%matplotlib inline

import numpy as np

import pandas as pd
# it allows you to quickly and easily visualize your data without having to save plots as image files or open them in a separate window
```

* **plot(x,y)**
* it help plot the x and y value with the help of the help of the default line style and color.
* **plot.axis([x_min,x_max,y_min,y_max])**
* it help scale the x-axis and y-axis from minimum(0,0 coordinate)to maximum values (max,max)
* **plot(x,y,color='colcr-name','marker-type',linestyle='line-style-name',linewidth=3,markersize=10)**
* here the x and y coordinates are changed from blue to color_name and can be marked using various shapes such as circles, diamonds, etc. these markers can be of varied sizes and widths.
* **plot.xlabel('x-axis')**
* Naming the x-axis with a label.
* **plot(x,y,label='sample line')**
* it help drow a sample line displayed as a legend.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.DataFrame({
    "Size" : [
        "s","m","l","xl","s","m","l","s","s","l","xl","m"
    ]
})
df
```

```python
ord_data = df["Size"].unique()
```

```python
# 1. Categories must be a list of lists
oe = OrdinalEncoder(categories=[ord_data]) 

# 2. Input to fit must be 2D (use double brackets)
oe.fit(df[["Size"]]) 
```

```python
df["Size_en"] = oe.transform(df[["Size"]])
```

```python
df
```

```python
ord_data1 = {
    "s" : 0,
    "m" : 1,
    "l" : 2, 
    "xl" : 3
}
df["Size_en_map"]=df["Size"].map(ord_data1)
df
```

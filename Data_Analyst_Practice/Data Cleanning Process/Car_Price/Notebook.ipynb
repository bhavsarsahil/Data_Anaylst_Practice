# Car Name

```python
data['company'] = data['car_name'].str.split().str[0]
```

```python
def car_name(x):
    return x[x.index(" ")+1:]
```

```python
data["car_name"] = data["car_name"].apply(car_name)
```

# car_price_rupees

35,000 , 28.02 lakh , 1.02 crore 

```python
data["car_prices_in_rupee"]
```

```python
data["car_prices_in_rupee"] = data["car_prices_in_rupee"].str.replace(",", "")
```

```python
data["car_prices_in_rupee"]
```

```python
def clean_price(x):
    x = str(x).lower().strip()
    x = x.replace(',', '')  # remove commas
    
    if 'lakh' in x:
        return float(x.replace('lakh', '').strip()) * 100000
    elif 'crore' in x:
        return float(x.replace('crore', '').strip()) * 10000000
    else:
        return float(x)
```

```python
data['car_prices_in_rupee'] = data['car_prices_in_rupee'].apply(clean_price)
```

```python
data['car_prices_in_rupee']
```

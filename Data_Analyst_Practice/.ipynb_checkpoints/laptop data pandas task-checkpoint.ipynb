1. Remove GB From 'Ram' Column.

2. Remove KG From 'Weight' column.

3. Change Datatype Of 'Ram' & 'Weight' columns on that Values.

4. Add New Column named 'Touchscreen' and put 1 Where Touchscreen value available else 0.

5. Add New Column named 'IPS' and put 1 Where IPS value available else 0.

6. Split Resoluation In Two Columns 'X_Res' & 'Y_Res'. ex.2560x1600 'X_Res' = 2560 , 'Y_Res' = 1600.

7. Add New Column named 'ppi' & Find ppi Value Of All Rows By X_Res,Y_Res,Inches.
    ppi =  (((X_Res)*2)+((Y_Res)2))*0.5) / Inches.

8. Add New Column named 'CPU Name' and add Value From 'cpu' column. ex. Intel Core i5

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("laptop_data.csv")
df
```

```python
# df["Ram"]=df["Ram"].replace("GB","",regex=True)
```

```python
df["Ram"]=df["Ram"].str.replace("GB","")
```

```python
df
```

```python
# df["Ram"].replace(['8GB', '16GB', '4GB', '2GB', '12GB', '6GB', '32GB', '24GB', '64GB'],["8","16","4","2","12","6","32","24","64"],inplace=True)
```

```python
df["Weight"].unique()
```

```python
df["Weight"]=df["Weight"].replace("kg","",regex=True)
```

```python
df
```

```python
df["Ram"]=df["Ram"].astype(int)
```

```python
df.info()
```

```python
df["Weight"]=df["Weight"].astype(float)
```

```python
# df["Touchscreen"]=df["ScreenResolution"].apply(lambda x:1 if "Touchscreen" in x else 0)
```

```python
df["Touchscreen"]=df["ScreenResolution"].str.contains("Touchscreen").astype(int)
```

```python
df
```

```python
df["IPS"]=df["ScreenResolution"].str.contains("IPS").astype(int)
```

```python
df
```

```python
new=df["ScreenResolution"].str.split("x",expand=True)
```

```python
df["X_RES"]=new[0]
df["Y_RES"]=new[1]
```

```python
df["X_RES"]=df["X_RES"].replace("[A-Za-z/|+]","",regex=True)
```

```python
df
```

```python
df["X_RES"]=df["X_RES"].replace(["4     3840","4   3840"],"3840",regex=True).astype(float)
```

```python
df["Y_RES"]=df["Y_RES"].astype(float)
```

```python
df["ppi"] = (((df["X_RES"]*2)+(df["Y_RES"]*2))*0.5) / df["Inches"]
```

```python
df
```

```python
df["Cpu"]=df["Cpu"]
df  
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
data = {
    'name' : ['sahil', "amit" , "parth" ,"yash", "kasim" ],
    'age' : [25, None, 44, 23, None],
    "salary" : [50000 , 60000, 70000, None, None],
    "gender" : ["male", "male" , "female", "male", "female"],
    "passed" : ["yes", "no" , "yes", "no","yes"]
    
}

df = pd.DataFrame(data)
df
# df.isnull().mean() * 100
le = LabelEncoder()

df["Gender_Encoded"] = le.fit_transform(df['gender'])
df["Passed_Encoded"] = le.fit_transform(df['passed']) 

print(df[["name", "gender", "Gender_Encoded", "passed", "Passed_Encoded"]])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np
```

```python
data = {
    "Days": [1, 2, 3, 4, 5],
    "NOP": [50,40,30,45,69]
}

df = pd.DataFrame(data)
print(df)

sns.lineplot(x="Days", y="NOP", data=df)
plt.title("NOP over Days")
plt.xlabel("Days")
plt.ylabel("NOP")
plt.show()
```

```python
plt.figure(figsize=(17,7))
sns.set_theme(style="darkgrid")
data = pd.read_excel("D:/GitHub Data_Analyst/Data_Anaylst_Practice/Data_Analyst_Practice/Pandas/ESD.xlsx")
# print(data.head(10)) 
sns.lineplot(data = data , x = "Business Unit", y = "Age", hue ="Ethnicity")

plt.title("Line Plot",fontsize=25)
plt.xlabel("Business Unit",fontsize=15)
plt.ylabel("Age",fontsize=15)
```

# Bar Plot In Seaborn

```python
data = sns.load_dataset("tips")
# print(data)
sns.barplot(x="day", y="total_bill", data=data, hue='sex' , color='g', order=["Sun", "Sat", "Fri", "Thur"], errorbar=("ci", 0))
plt.show()  
```

# Histogram 

```python
sns.histplot(data=data, x="day", hue='sex', kde=True, edgecolor='black')
```

```python
plt.figure(figsize=(14,7))
sns.set_theme(style="darkgrid") 
data = sns.load_dataset("titanic")
sns.histplot(data=data, x="age", hue='who', kde=True, edgecolor='black', bins=30)
```

```python
data = sns.load_dataset("tips")
plt.figure(figsize=(10,7))
sns.scatterplot(x="total_bill", y="tip", data=data, hue='time', size="size", edgecolor='black', palette="Set2", markers="o", sizes=(50, 200))
plt.title("Scatter Plot", fontsize=25)  
plt.xlabel("Total Bill", fontsize=15)
plt.ylabel("Tip", fontsize=15)
plt.legend(title='Day of Week', bbox_to_anchor=(1, 1))
plt.show()
```

# Count Plot

```python
sns.countplot(x="day", data=data, palette="Set2", edgecolor='black')
plt.title("Count Plot", fontsize=25)
plt.xlabel("Sex", fontsize=15)
plt.ylabel("Count", fontsize=15)    
plt.legend(title='Day of Week', bbox_to_anchor=(1, 1))
plt.show()
```

## Date & Time Function with SQL database 

```python
df_order = pd.read_sql_query('SELECT * FROM orders', conn)

df_order['orderDate'] = pd.to_datetime(df_order['orderDate'])
df_order['shippedDate'] = pd.to_datetime(df_order['shippedDate'])
df_order['DeliveryTime'] = (df_order['shippedDate'] - df_order['orderDate']).dt.days
result=df_order[['orderDate', 'shippedDate', 'DeliveryTime']]
result
```

## Quick Cheat Sheet for Date Parts:
| Goal	  | SSMS (T-SQL) | 	Pandas |
|---|---|---|
| Day Name  | 	DATENAME(dw, date) | 	.dt.day_name() |
| Month Name| DATENAME(mm, date) | .dt.month_name() |
| Day of Month | 	DAY(date) | .dt.day |
| Year| 	YEAR(date)| 	.dt.year |

## Case Function

```python
df_product = pd.read_sql_query("select * from products", conn)
```

```python
df_product['Production_details'] = np.where(
    df_product['quantityInStock'] < 1000,
    "Urgent need of Stock",
    "No Requirement"
)

result = df_product[['productName','quantityInStock','Production_details']]
result
```

# Group By Function 

```python
result = df.groupby('department')['employee_id'].count()
result
```

```python
result = df.groupby('department')['salary'].sum().round(1)
result
```

```python
result = df_product.groupby('productLine')['quantityInStock'].sum().sort_values(ascending=False)
result
```

# Having Clause 

```python
# 1. Group and Sum (SELECT ... GROUP BY)
result = df_product.groupby('productLine', as_index=False).agg(Stock=('quantityInStock', 'sum'))

# 2. Filter the result (HAVING SUM > 20000)
result = result[result['Stock'] > 20000]

# 3. Sort the values (ORDER BY Stock DESC)
result = result.sort_values(by='Stock', ascending=False)

# Display result
print(result)
```

# Joins In Pandas using Sql File

```python
df_Customers = pd.read_sql_query("Select * from Customers", conn)
df_Local_Orders = pd.read_sql_query("Select * from Local_Orders", conn)
```

```python
df_merge = pd.merge(df_Customers,df_Local_Orders, how='inner', on='CustomerID')
df_merge
```

### Left Join

```python
df_merge = pd.merge(df_Customers,df_Local_Orders, how='left', on='CustomerID')
df_merge
```

### Right Join

```python
df_merge = pd.merge(df_Customers,df_Local_Orders, how='right', on='CustomerID')
df_merge
```

### Cross Join

```python
df_merge = pd.merge(df_Customers,df_Local_Orders, how='cross')
df_merge
```

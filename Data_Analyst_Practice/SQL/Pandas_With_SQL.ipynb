## Date & Time Function with SQL database 

```python
df_order = pd.read_sql_query('SELECT * FROM orders', conn)

df_order['orderDate'] = pd.to_datetime(df_order['orderDate'])
df_order['shippedDate'] = pd.to_datetime(df_order['shippedDate'])
df_order['DeliveryTime'] = (df_order['shippedDate'] - df_order['orderDate']).dt.days
result=df_order[['orderDate', 'shippedDate', 'DeliveryTime']]
result
```

## Quick Cheat Sheet for Date Parts:
| Goal	  | SSMS (T-SQL) | 	Pandas |
|---|---|---|
| Day Name  | 	DATENAME(dw, date) | 	.dt.day_name() |
| Month Name| DATENAME(mm, date) | .dt.month_name() |
| Day of Month | 	DAY(date) | .dt.day |
| Year| 	YEAR(date)| 	.dt.year |
